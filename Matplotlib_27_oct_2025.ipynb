```python
import matplotlib.pyplot as plt
import numpy as np

# Create range of x values
x = np.linspace(0, 10, 100)

# Define multiple mathematical functions
y1 = x                  # Linear function y = x
y2 = x**2               # Quadratic function y = x²
y3 = np.sqrt(x)         # Square root function y = √x

# Create figure
plt.figure(figsize=(8, 5))

# Plot multiple lines with different styles
plt.plot(x, y1, label='y = x', color='blue', linestyle='-')
plt.plot(x, y2, label='y = x²', color='green', linestyle='--')
plt.plot(x, y3, label='y = √x', color='red', linestyle=':')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Line Graph: Linear, Quadratic, and Square Root Functions')

# Add legend and grid
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```
