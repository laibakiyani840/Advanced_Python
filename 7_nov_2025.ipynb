##### **Sharing Data and Behaviour Across All Object**

```python
class Employee:
    company_name = "TechCorp"  # Class variable (shared by all instances)

    def __init__(self, name):
        self.name = name

    @staticmethod
    def company_policy():
        print("All employees must adhere to the company's code of conduct.")
```

```python

emp1 = Employee("Ali")
emp2 = Employee("Laiba")
print(emp1.name, "works at", Employee.company_name)
Employee.company_policy()
```

```python
class Student:
    school_name = "Greenwood High"  # Class variable/static variable

    def __init__(self, name):
        self.name = name # Instance variable
s1 = Student("Ahmed")  # s1 & s2 are objects/instances of Student class
s2 = Student("Sara")
```

```python
print(s1.school_name)
```

```python
Student.school_name = "Sunrise Academy"  # Modifying class variable
print(s2.school_name)  # Outputs: Sunrise Academy
```
